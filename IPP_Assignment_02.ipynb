## 1. Read and display an image using OpenCV

**Aim:** Read `flower.jpg` and display it in a window titled "Original Image".

### Explanation
1. `cv2.imread()` reads the image file.
2. `img is None` checks whether loading failed.
3. `cv2.imshow()` displays the image with the specified title.
4. `cv2.waitKey(0)` keeps the window open until a key is pressed.
5. `cv2.destroyAllWindows()` closes OpenCV windows.

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("flower.jpg")

# Check if the image was loaded
if img is None:
    print("Image not found! Check the filename and folder.")
else:
    # Convert BGR to RGB
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Display the image inside Jupyter Notebook
    plt.imshow(img)
    plt.title("Original Image")
    plt.axis("off")
    plt.show()
```

## 2. Print image dimensions and data type

**Aim:** Print the height, width, number of channels, and data type of `scene.jpg`.

### Explanation
1. `img.shape[:2]` gives height and width.
2. `img.ndim` tells us how many dimensions the image array has.
3. `img.shape[2]` gives the number of channels for a colour image.
4. `img.dtype` reports the pixel data type, usually `uint8`.

```python
import cv2

# Read the image
img = cv2.imread("scene.jpg")

# Check if the image exists
if img is None:
    print("Image not found! Check the filename and folder.")
else:
    # Get image height and width
    height, width = img.shape[:2]

    # Get the number of colour channels
    channels = 1 if img.ndim == 2 else img.shape[2]

    # Print image information
    print("Height:", height)
    print("Width:", width)
    print("Channels:", channels)
    print("Data type:", img.dtype)
```

## 5. Resize an image to 50% of its original width and height

**Aim:** Resize `flower.jpg` so its width and height are each half of the original dimensions.

### Explanation
1. `img.shape[:2]` retrieves the original height and width.
2. Multiplying each dimension by `0.5` makes it half-sized.
3. `cv2.resize()` creates the resized image.
4. `cv2.imshow()` displays it; `cv2.imwrite()` also saves a copy.

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("flower.jpg")

if img is None:
    print("Image not found! Check the filename and folder.")
else:
    # Get the original height and width
    height, width = img.shape[:2]

    # Calculate the new dimensions (50% size)
    new_width = int(width * 0.5)
    new_height = int(height * 0.5)

    # Resize the image
    resized = cv2.resize(img, (new_width, new_height))

    # Save the resized image
    cv2.imwrite("resized_flower.jpg", resized)
    print("Resized image saved as resized_flower.jpg")

    # Convert BGR to RGB for correct colours
    resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)

    # Display inside Jupyter Notebook
    plt.imshow(resized_rgb)
    plt.title("Resized Image (50%)")
    plt.axis("off")
    plt.show()
```

## 6. Crop the central 200 × 200 pixel area using Matplotlib

**Aim:** Crop the central 200 × 200 pixel region of `flower.jpg` and display it using Matplotlib.

### Explanation
1. The centre is found by dividing width and height by two.
2. The crop starts 100 pixels before the centre and ends 100 pixels after it.
3. `img[y1:y2, x1:x2]` crops rows first, then columns.
4. BGR is converted to RGB so Matplotlib displays colours correctly.

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("flower.jpg")

if img is None:
    print("Image not found! Check the filename and folder.")
else:
    # Get the image dimensions
    height, width = img.shape[:2]

    if height < 200 or width < 200:
        print("The image must be at least 200 x 200 pixels.")
    else:
        # Calculate the centre coordinates
        center_x = width // 2
        center_y = height // 2

        # Calculate the crop boundaries
        x1 = center_x - 100
        y1 = center_y - 100
        x2 = center_x + 100
        y2 = center_y + 100

        # Crop the central 200 x 200 region
        cropped = img[y1:y2, x1:x2]

        # Convert BGR to RGB for Matplotlib
        cropped_rgb = cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB)

        # Display the cropped image
        plt.imshow(cropped_rgb)
        plt.title("Central 200 x 200 Crop")
        plt.axis("off")
        plt.show()
```

## 7. Rotate an image 90 degrees clockwise

**Aim:** Rotate `flower.jpg` 90 degrees clockwise using OpenCV.

### Explanation
1. `cv2.rotate()` rotates the image.
2. `cv2.ROTATE_90_CLOCKWISE` specifies a 90-degree clockwise rotation.
3. `cv2.imshow()` displays the result and `cv2.imwrite()` saves it.

**Expected output:** The image displayed rotated clockwise and saved as `rotated_flower.jpg`.

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("flower.jpg")

if img is None:
    print("Image not found! Check the filename and folder.")
else:
    # Rotate the image 90 degrees clockwise
    rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

    # Save the rotated image
    cv2.imwrite("rotated_flower.jpg", rotated)
    print("Rotated image saved as rotated_flower.jpg")

    # Convert BGR to RGB for correct colours
    rotated_rgb = cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)

    # Display the rotated image inside Jupyter
    plt.imshow(rotated_rgb)
    plt.title("Rotated Image (90° Clockwise)")
    plt.axis("off")
    plt.show()
```

## 9. Convert RGB to HSV and display each channel

**Aim:** Convert `flower.jpg` from RGB to HSV and display Hue, Saturation, and Value separately using Matplotlib.

### Explanation
1. OpenCV reads colour images in BGR order, so the code first converts BGR to RGB.
2. `cv2.COLOR_RGB2HSV` converts RGB to HSV.
3. The third array index selects the channel: `0` for Hue, `1` for Saturation, `2` for Value.
4. Matplotlib displays each channel in grayscale.


```python
import cv2
import matplotlib.pyplot as plt

# Read the image (OpenCV loads it in BGR order)
img = cv2.imread("flower.jpg")

if img is None:
    print("Image not found! Check the filename and folder.")
else:
    # Convert BGR to RGB
    rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Convert RGB to HSV
    hsv = cv2.cvtColor(rgb, cv2.COLOR_RGB2HSV)

    # Extract the H, S and V channels
    hue = hsv[:, :, 0]
    saturation = hsv[:, :, 1]
    value = hsv[:, :, 2]

    # Create a figure with three side-by-side plots
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(hue, cmap="gray")
    plt.title("Hue")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(saturation, cmap="gray")
    plt.title("Saturation")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(value, cmap="gray")
    plt.title("Value")
    plt.axis("off")

    # Display all three channels
    plt.tight_layout()
    plt.show()
```

## 10. Calculate Mean Squared Error (MSE) between two grayscale images

**Aim:** Calculate the MSE between `img1.jpg` and `img2.jpg` after loading them as grayscale images.

### Explanation
1. Both images are loaded in grayscale.
2. The code checks that both images exist and have matching dimensions.
3. `astype(np.float64)` supports precise subtraction and squaring.
4. `(img1 - img2) ** 2` calculates squared pixel differences.
5. `np.mean()` averages all squared differences to produce the MSE.

**Expected output:** A numeric MSE value. MSE is 0 when both images have identical pixel values; larger values indicate greater pixel-level differences.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read both images in grayscale
img1 = cv2.imread("img1.jpg", cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("img2.jpg", cv2.IMREAD_GRAYSCALE)

if img1 is None or img2 is None:
    print("One or both images not found!")

else:
    # Resize Image 2 to match Image 1
    img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

    # Display both images inside Jupyter
    plt.figure(figsize=(8, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(img1, cmap="gray")
    plt.title("Image 1")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(img2, cmap="gray")
    plt.title("Image 2 (Resized)")
    plt.axis("off")

    plt.show()

    # Convert pixel values to float
    img1 = img1.astype(np.float64)
    img2 = img2.astype(np.float64)

    # Calculate Mean Squared Error
    mse = np.mean((img1 - img2) ** 2)

    print("Mean Squared Error:", mse)
```
